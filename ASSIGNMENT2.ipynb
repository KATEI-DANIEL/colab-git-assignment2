```python
from getpass import getpass
import os

github_username = "katei-daniel"
github_token = getpass("Enter your GitHub PAT: ")
```

```python
os.environ["GITHUB_USERNAME"] = github_username
os.environ["GITHUB_TOKEN"] = github_token

with open("/tmp/askpass.sh", "w") as f:
    f.write("""#!/bin/sh
case "$1" in
  *Username*) echo "$GITHUB_USERNAME" ;;
  *Password*) echo "$GITHUB_TOKEN" ;;
esac
""")
os.chmod("/tmp/askpass.sh", 0o700)

os.environ["GIT_ASKPASS"] = "/tmp/askpass.sh"
os.environ["GIT_TERMINAL_PROMPT"] = "0"
```
